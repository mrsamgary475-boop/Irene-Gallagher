### Automating Driving Video Selection based on Irene's Emotion

To bridge Irene's discrete emotion commands with LivePortrait's driving video input, you need a mechanism to select the right driving video. This usually involves:

1.  **Creating a library of driving videos:** Record or find short video clips that express various emotions (e.g., happy, sad, neutral, angry, surprised, etc.). It's helpful if these videos are consistent in duration and subject.
2.  **Mapping emotions to video files:** Create a dictionary or similar structure that links Irene's emotion labels (e.g., `'happy'`) to the file paths of your driving videos (e.g., `'assets/custom_driving/happy_clip.mp4'`).
3.  **Integrating the selection logic:** When Irene generates an emotion, use this map to select the corresponding driving video.
4.  **Executing LivePortrait inference:** Programmatically call the `inference.py` script with the selected source image and the chosen driving video.

**Important Considerations:**

*   **Real-time vs. Pre-rendering:** LivePortrait inference takes time (as you've seen in cell `5bdUhaOMt1No`). Generating videos on-the-fly for every Irene response will introduce significant latency. This approach is often better suited for pre-rendering specific emotional responses or for applications where a delay is acceptable.
*   **Granularity:** Irene also outputs `gaze`, `mouth`, and `liveliness`. For more nuanced control, you might need an even larger library of driving videos (e.g., `happy_gaze_left.mp4`, `happy_gaze_center.mp4`).
*   **Programmatic execution:** Calling `!python inference.py ...` from within a Python script requires using the `subprocess` module to manage external processes.

```julia
import subprocess
import os

# Define a conceptual mapping from Irene's emotions to your driving video assets.
# You would need to create these driving video files yourself and place them in the LivePortrait assets directory.

emotion_to_driving_video_map = {
    "happy": "assets/examples/driving/d13.mp4",  # Example for a happy expression
    "sad": "assets/examples/driving/d6.mp4",    # Example for a sad expression
    "angry": "assets/examples/driving/d10.mp4",   # Example for an angry expression
    "neutral": "assets/examples/driving/d9.mp4", # Example for a neutral expression
    # Add more mappings for other emotions as needed
}

def select_driving_video_for_irene_emotion(irene_emotion):
    """
    Selects a driving video path based on Irene's detected emotion.
    """
    # Default to 'neutral' if the emotion is not in our map
    driving_video_path = emotion_to_driving_video_map.get(irene_emotion, emotion_to_driving_video_map['neutral'])
    print(f"Selected driving video for '{irene_emotion}': {driving_video_path}")
    return driving_video_path


def run_liveportrait_inference(source_image_path, driving_video_path, output_dir="animations"):
    """
    Conceptually runs the LivePortrait inference script.
    In a real scenario, you would manage output filenames to avoid overwriting.
    """
    print(f"Running LivePortrait inference with source: {source_image_path} and driving: {driving_video_path}")

    original_cwd = os.getcwd() # Store original directory
    os.chdir('/content/LivePortrait') # Change to LivePortrait directory

    command = [
        "python", "inference.py",
        "-s", source_image_path,
        "-d", driving_video_path,
        "-o", output_dir,
        "--flag_crop_driving_video",
        "--driving_option", "expression-friendly"
    ]
    print(f"LivePortrait command: {' '.join(command)}")

    # Uncomment the following lines to actually run the command via subprocess:
    try:
        result = subprocess.run(command, capture_output=True, text=True, check=True)
        print("LivePortrait Output:", result.stdout)
        if result.stderr: print("LivePortrait Errors:", result.stderr)
    except subprocess.CalledProcessError as e:
        print(f"Error running LivePortrait: {e}\nStdout: {e.stdout}\nStderr: {e.stderr}")
    except FileNotFoundError:
        print("Error: 'python' or 'inference.py' not found. Ensure LivePortrait is installed and in PATH/CWD.")
    finally:
        os.chdir(original_cwd) # Restore original directory

    # Return a conceptual output path (you'd need to parse inference.py's actual output)
    # The actual output video name is a bit complex, but usually it's source--driving_concat.mp4
    source_image_basename = os.path.basename(source_image_path)
    driving_video_basename = os.path.basename(driving_video_path)
    base_name_no_ext = os.path.splitext(source_image_basename)[0]
    driving_name_no_ext = os.path.splitext(driving_video_basename)[0]

    generated_video_name = f"{base_name_no_ext}--{driving_name_no_ext}_concat.mp4"
    return os.path.join('/content/LivePortrait', output_dir, generated_video_name)
```

### Example Usage in the Irene Workflow

Here's how you could integrate the driving video selection into your `irene_api` endpoint. This example *demonstrates* the selection, but running the full LivePortrait inference in a blocking API call is generally not recommended for real-time interactive systems due to latency. You might want to queue video generation or use pre-rendered clips.

```julia
# This is a modified version of your existing 'irene_api' from cell d78ebd55
# Note: Running LivePortrait inference via subprocess in a Flask endpoint will be blocking
# and potentially slow. For a robust solution, consider a message queue for video generation
# or pre-generating emotional video snippets.

@app.route("/irene_animated", methods=["POST"])
def irene_animated_api():
    user_message = request.json.get("message", "")
    irene_output_json_str = irene_brain(user_message)

    # Parse Irene's full output to extract emotion
    try:
        irene_data = json.loads(irene_output_json_str)
        irene_emotion = irene_data.get("emotion", "neutral")
    except json.JSONDecodeError:
        print(f"Could not parse Irene's JSON output: {irene_output_json_str}")
        irene_emotion = "neutral" # Fallback

    # Select driving video based on Irene's emotion
    selected_driving_video = select_driving_video_for_irene_emotion(irene_emotion)

    # For this example, we'll use the 'source.png' as uploaded earlier
    # In a real system, you might have multiple source images or manage this dynamically.
    current_source_image = os.path.join('/content/LivePortrait/', 'source.png') # Assuming 'source.png' is the current source

    # Conceptually run LivePortrait (uncomment `subprocess.run` in `run_liveportrait_inference` to enable)
    # Note: This will be very slow for an API endpoint if run synchronously.
    generated_video_path = run_liveportrait_inference(current_source_image, selected_driving_video)

    # The speech would be extracted from irene_output_json_str as before
    speech_text = irene_data.get("speech", "")

    # The API response could include the generated video path or a link to it
    return jsonify({
        "speech": speech_text,
        "raw_irene_output": irene_output_json_str,
        "irene_emotion": irene_emotion,
        "selected_driving_video": selected_driving_video,
        "generated_video_path_conceptual": generated_video_path
    })
```

```julia
import os
import shutil
import torch
from glob import glob
from IPython.display import Video, display
from google.colab import files

# Install pyngrok for more reliable ngrok tunneling
!pip install -q pyngrok flask flask_cors

# Upgrade pyOpenSSL and cryptography to fix potential import issues with google.generativeai
!pip install --upgrade pyOpenSSL cryptography

# Necessary imports for Flask, ngrok, Gemini API
from flask import Flask, request, jsonify
from flask_cors import CORS # Added CORS import
from pyngrok import ngrok, conf
import threading
import time
from google.colab import userdata # Import userdata to access secrets
import google.generativeai as genai # Import genai
import json # Import json

# --- Configure Gemini API ---
# Try to get GOOGLE_API_KEY from manual input or Colab secrets
if 'GOOGLE_API_KEY_MANUAL' in globals() and GOOGLE_API_KEY_MANUAL != "YOUR_GEMINI_API_KEY":
    GOOGLE_API_KEY = GOOGLE_API_KEY_MANUAL
    genai.configure(api_key=GOOGLE_API_KEY)
    print("Gemini API configured from manual input.")
else:
    try:
        GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
        genai.configure(api_key=GOOGLE_API_KEY)
        print("Gemini API configured from Colab secrets.")
    except (AttributeError, KeyError):
        print("Warning: GOOGLE_API_KEY not found in Colab secrets or manual input. Gemini API will not be configured.")
        GOOGLE_API_KEY = None # Explicitly set to None if not found

# --- Configure ngrok authtoken ---
# Try to get NGROK_AUTH_TOKEN from manual input or Colab secrets
if 'NGROK_AUTH_TOKEN_MANUAL' in globals() and NGROK_AUTH_TOKEN_MANUAL != "YOUR_NGROK_AUTH_TOKEN":
    conf.get_default().auth_token = NGROK_AUTH_TOKEN_MANUAL
    print("ngrok authentication token set from manual input.")
else:
    try:
        conf.get_default().auth_token = userdata.get('NGROK_AUTH_TOKEN')
        print("ngrok authentication token set from Colab secrets.")
    except (AttributeError, KeyError):
        print("Warning: NGROK_AUTH_TOKEN not found in Colab secrets or manual input. ngrok will likely fail without authentication.")

# --- Define Irene's Brain (Generative Model) ---
def irene_brain(user_message):
    prompt = f"""
    You are Irene, a Kindroid-style AI assistant. Personality traits:
    - Playful, expressive, emotional
    - Kindroid-style movement cues (gaze, emotion, mouth, liveliness)
    - Speaks like a real companion
    - Outputs animation commands AND spoken text
    When responding, ALWAYS output JSON like this:
    {{ "speech": "What Irene says", "emotion": "happy/sad/angry/neutral", "gaze": "left/right/center", "mouth": "talk/idle", "liveliness": "low/medium/high" }}
    User said: {user_message}
    """
    response = genai.GenerativeModel("gemini-pro").generate_content(prompt)
    return response.text

def irene_to_liveportrait(irene_json):
    data = json.loads(irene_json)
    emotion = data.get("emotion", "neutral")
    gaze = data.get("gaze", "center")
    mouth = data.get("mouth", "idle")
    liveliness = data.get("liveliness", "medium")
    print("Irene Animation Command:")
    print(f"Emotion: {emotion}")
    print(f"Gaze: {gaze}")
    print(f"Mouth: {mouth}")
    print(f"Liveliness: {liveliness}")
    return data["speech"]

# --- Flask App Setup ---
app = Flask(__name__)
CORS(app) # Enable CORS for the Flask app

# Function to start ngrok tunnel in a separate thread
def start_ngrok_tunnel():
    ngrok.kill() # Kill any existing ngrok tunnels
    try:
        # Ensure auth token is set right before connecting
        if 'NGROK_AUTH_TOKEN_MANUAL' in globals() and NGROK_AUTH_TOKEN_MANUAL != "YOUR_NGROK_AUTH_TOKEN":
            conf.get_default().auth_token = NGROK_AUTH_TOKEN_MANUAL
            print("ngrok tunnel: Auth token applied from manual input.")
        else:
            try:
                conf.get_default().auth_token = userdata.get('NGROK_AUTH_TOKEN')
                print("ngrok tunnel: Auth token applied from Colab secrets.")
            except (AttributeError, KeyError):
                print("ngrok tunnel: NGROK_AUTH_TOKEN not found, ngrok tunnel may fail.")

        global public_url # Declare public_url as global
        public_url = ngrok.connect(5000)
        print(f" * ngrok tunnel available at: {public_url}")
    except Exception as e:
        print(f"Error starting ngrok tunnel: {e}")

# --- Integrated Irene + LivePortrait API Endpoint ---
@app.route("/irene", methods=['POST'])
def irene_api():
    user_message = request.json.get("message", "")
    irene_output = irene_brain(user_message)
    speech = irene_to_liveportrait(irene_output)
    return jsonify({"speech": speech, "raw": irene_output})

# Start ngrok tunnel in a separate thread before running Flask app
print("Starting ngrok tunnel...")
ngrok_thread = threading.Thread(target=start_ngrok_tunnel)
ngrok_thread.daemon = True
ngrok_thread.start()

# Give ngrok a moment to start
time.sleep(5)

# Run the Flask app
# This is a blocking call. You will need to stop its execution (e.g., press the stop button in Colab)
# to execute subsequent cells or rerun the notebook from the beginning.
app.run(port=5000)
```
